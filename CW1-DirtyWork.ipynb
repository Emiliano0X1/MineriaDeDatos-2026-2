```python
import random

random_list = [random.random() for _ in range(5_000_000)]

print("Random List Length : ", len(random_list))

```

```python
def st_derivation(random_list):
    average = 0
    tempSum = 0
    n = len(random_list)
    for num in random_list:
        tempSum += num

    average = tempSum / n


    above_ope = 0
    for num in random_list:
        above_ope += pow(num - average, 2)

    newTemp = above_ope / n
    res = pow(newTemp, 0.5)
    return res
```

```python
print('Std: ', {st_derivation(random_list)})
```

```python
import numpy as np

np_array = np.array(random_list)

print('Std: ', {np_array.std()})
```
