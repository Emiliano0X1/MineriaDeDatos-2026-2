```python
import pandas as pd
import numpy as np 
 
url = "https://raw.githubusercontent.com/mricardo89/data-mining/main/Unit-II/datasets/titanic.csv"
df = pd.read_csv(url)
```

Asimetría (Skewness): Investiguen cómo aplicar la función .skew() en la columna Age. Si el resultado es positivo o negativo, ¿qué significa físicamente sobre la edad de los pasajeros en el barco? (¿Había más jóvenes o más ancianos?).

Al ser un sesgo positivo, lo que nos dice es que habia una gran cocnentracion de valores menores ( Osea jovenes entre 18 a 35 años) pero habia valores muy altos o perosnas con edades muy avanzadas que hacian que la media fuera sueoprioir a la mediana, teninedo este efecto como resultado

```python
print("La asimetria de las edades es de :",df['Age'].skew())
```

Curtosis (Kurtosis): Investiguen cómo aplicar la función .kurt() en la columna Fare. ¿Qué significa este número sobre el "grosor" de las colas de la distribución? ¿Confirma esto la existencia de los outliers que encontramos hoy en clase?


```python
print("La curtosis fue de :" ,df['Fare'].kurt())
```

Tablas Cruzadas (Cross-Tabulation): Investiguen la función pd.crosstab(). Generen una tabla cruzada entre la clase del pasajero (Pclass) y si sobrevivió (Survived). Esta técnica de sumarización matricial es el equivalente a una Tabla Dinámica en Excel. ¿Qué patrón logran visualizar rápidamente al cruzar ambas variables?

Lo que pude notar es que la mayoria de las personas que no sobrevivieron pertenecian a la clase mas baja posible, mientras que un gran porcen traje de los sobrevivientes vino de la clase 1, aunque eso si habia mucha mas gente en la clase 3 pero al proprocion es muy alta haciendo que casi solo una tercera parte sobreviviera, en la segunda un 50/50 y en la primera mas de la mitad


```python
newTable = pd.crosstab(df['Pclass'], df['Survived'], margins=True, margins_name="Total")
print(newTable)
```
