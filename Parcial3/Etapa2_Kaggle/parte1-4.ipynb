```python
import pandas as pd
import numpy as np

df_train = pd.read_csv('train_p2.csv')
df_test = pd.read_csv('test_p2.csv')

##1. Jerarquia 

def convertData(df):

    ##1. Jerarquia 
    map_nivel_verificacion = {'NINGUNO' : 1, 'CORREO' : 2, 'TELEFONO' : 3, 'VERIFICADO' : 4}
    df['nivel_verificacion'] = df['nivel_verificacion'].map(map_nivel_verificacion)

    ##2. Etiquetas sin jerarquia
    df = pd.get_dummies(df, columns = ['pais', 'dispositivo', 'email_domain'], dtype= int)

    # 3. Trigonometrico
    alpha = (df['hora_pico_actividad'] - 1) * 2 * np.pi / 12
    df['hora_pico_actividad_sin'] = np.sin(alpha)
    df['hora_pico_actividad_cos'] = np.cos(alpha)

    ##Dias
    fecha = pd.to_datetime(df['fecha_creacion'], format='%Y-%m-%d')
    today = pd.to_datetime('2026-11-25', format='%Y-%m-%d')

    ratio_dias = (today - fecha).dt.days
    df['fecha_creacion'] = ratio_dias

    #4. Miles de datos
    df['seguidores'] = np.log1p(df['seguidores'])
    df['siguiendo'] = np.log1p(df['siguiendo'])
    df['posts_totales'] = np.log1p(df['posts_totales'])
    df['likes_given_30d'] = np.log1p(df['likes_given_30d'])

    

    return df

df_train = convertData(df_train)
df_train_y = df_train['is_bot']
df_train = df_train.drop(columns=['is_bot', 'nivel_verificacion', 'user_id'])
df_test = convertData(df_test)
df_test_ids = df_test['user_id']
df_test = df_test.drop(columns=['nivel_verificacion', 'user_id'])



from caja_negra_p2 import evaluar_y_generar_submission

cv_f1, df_sub = evaluar_y_generar_submission(
    X_train=df_train,
    y_train=df_train_y,
    X_test=df_test,
    test_user_ids=df_test_ids,
    output_path="submission_p2.csv"
)

```
