```python
import pandas as pd
url = "https://raw.githubusercontent.com/mricardo89/data-mining/main/Unit-II/datasets/googleplaystore.csv"
df = pd.read_csv(url)
```

Pregunta A: El scraper cometió errores y guardó la misma app varias veces. Escriban una línea de código para identificar si hay nombres duplicados en la columna App (df.duplicated(subset=['App']).sum()). Si los hay, apliquen .drop_duplicates(subset=['App'], keep='first', inplace=True) para quedarse solo con la primera aparición de cada aplicación

```python
res = df.duplicated(subset=['App']).sum()
print("Cantidad de duplicados: ", res)
```

```python
## Si hay registros duplicados
df.drop_duplicates(subset=['App'], keep='first', inplace=True)

res = df.duplicated(subset=['App']).sum()
print("Cantidad de duplicados: ", res)
```

Pregunta B: La columna Installs tiene texto como "1,000,000+". Un algoritmo no puede procesar eso.

a. Escriban una instrucción encadenando el método .str.replace() para eliminar el símbolo + y otro para eliminar la coma ,. (Ojo: utilicen regex=False o escapen el símbolo más \+).

```python
df['Installs'] = df['Installs'].str.replace('+',"",regex=False)
df['Installs'] = df['Installs'].str.replace(',',"",regex=False)

df['Installs'] = pd.to_numeric(df['Installs'],errors = 'coerce') ##There some texts like Free that generate erros, so i passed as NaN cause theres no any value

print(df['Installs'])
```

b. Al final de la cadena, utilicen .astype(int) (o float si tienen NaNs) para convertir la columna.

```python
df['Installs'] = df['Installs'].astype(float)
print(df['Installs'].dtype)
```

Comprueben el éxito ejecutando df['Installs'].mean(). 
¿Cuál es el promedio real de descargas?

El promedio de las descargas son de 7,777,506.73 

```python
average = df['Installs'].mean()
print("Downloads Average: ", average)
```

Pregunta C: La columna Price tiene valores como "$4.99" y "0". Apliquen la misma lógica arquitectónica que en la Pregunta B: eliminen el símbolo de dólar $ y conviertan la columna usando .astype(float).

```python
##Same process - delete $ sign and convert to int or float
df['Price'] = df['Price'].str.replace('$','',regex=False)
df['Price'] = df['Price'].str.replace('.','',regex=False)

df['Price'] = pd.to_numeric(df['Price'], errors='coerce')

df['Price'] = df['Price'].astype(float)

print(df['Price'].dtype)
```
